```python
import json
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM, BitsAndBytesConfig
import re
```

```python
model_name = "qwen/qwen-7b"
tokenizer = AutoTokenizer.from_pretrained(model_name, trust_remote_code=True)

# Sử dụng device_map='auto' và offload để chuyển layers không cần thiết sang CPU
# Sử dụng thư mục offload trên Colab: '/content/offload'
model = AutoModelForCausalLM.from_pretrained(
    model_name,
    trust_remote_code=True,
    torch_dtype=torch.float16,
    device_map="auto",
    offload_folder="/content/offload",
    offload_state_dict=True
)
model.eval()
```

```python
def label_durationqa(sample, max_new_tokens=128):
    """Trả về list các 'yes'/'no' ứng với sample, kèm ví dụ minh họa và điều kiện nhãn."""
    # Ví dụ để mô hình hiểu rõ hơn
    example = {
        "context": "Tôi đang sửa chữa chiếc xe đạp bị hỏng.",
        "options": ["30 phút", "1 tháng", "10 phút", "2 giờ"],
        "qid": 54,
        "question": "Mất thời gian bao lâu để sửa chữa chiếc xe đạp?",
        "labels": ["yes", "no", "yes", "yes"],
    }

    # Xây dựng prompt: có ví dụ và điều kiện
    prompt = (
        "Example:\n"
        + json.dumps(example, ensure_ascii=False, indent=2)
        + "\n\n"
        f"Context: {sample['context']}\n"
        f"Question: {sample['question']}\n"
        "Options:\n"
    )
    for idx, opt in enumerate(sample["options"], 1):
        prompt += f"  {idx}. {opt}\n"

    # Điều kiện: ít nhất phải có một 'yes' hoặc 'no' trong nhãn
    prompt += (
        "\nPlease answer for each option with 'yes' or 'no', "
        "in the format: labels: [\"yes\", \"no\", ...]. "
        "Ensure the labels list contains at least one 'yes' or 'no'.\n"
    )

    # Giải phóng bộ nhớ GPU trước khi generate
    if torch.cuda.is_available():
        torch.cuda.empty_cache()
    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)
    with torch.no_grad():
        out = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            do_sample=False,
            use_cache=False,
            eos_token_id=tokenizer.eos_token_id,
            pad_token_id=tokenizer.pad_token_id
        )
    text = tokenizer.decode(out[0], skip_special_tokens=True)
    generated = text[len(prompt):].strip()
    m = re.search(r'labels\s*[:=]\s*(\[[^\]]*\])', generated)
    if m:
        labels = json.loads(m.group(1).replace("'", '"'))
    else:
        toks = re.findall(r'\byes\b|\bno\b', generated.lower())
        labels = toks[:len(sample["options"])]
    return labels
```

```python
def main():
    input_path = "/content/drive/MyDrive/AI_Course/VLSP/public_test.txt"
    output_path = "/content/drive/MyDrive/AI_Course/VLSP/QWEN_public_test.txt"

    with open(input_path, "r", encoding="utf-8") as fin, \
         open(output_path, "w", encoding="utf-8") as fout:

        for line in fin:
            line = line.strip()
            if not line:
                continue
            sample = json.loads(line)
            sample["labels"] = label_durationqa(sample)
            fout.write(json.dumps(sample, ensure_ascii=False) + "\n")

    print(f"Done! Kết quả đã được ghi vào {output_path}")
```

```python
if __name__ == "__main__":
    main()
```
